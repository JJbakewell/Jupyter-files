## [2] Index and slicing

```python
name_list = ["Leo", "Nora", "Emma", "James", "Lucy"]
```

```python
print(name_list[1])
```

```python
print(name_list[0:2])
```

```python
#If you omit the first index, the slice starts at the beginning. 
print(name_list[:2])
```

```python
#If you omit the second, the slice goes to the end. 
print(name_list[3:])
```

```python
#If you omit both, the slice is a copy of the whole list.
print(name_list[:])
```

- **Nagative index**

```python
name_list = ["Leo", "Nora", "Emma", "James", "Lucy"]
print(name_list[-1])
```

```python
print(name_list[-3:-1])
```

```python
print(name_list[-3:])
```

- **A list of lists**

```python
#[day,month]
birthday_ddmm = [[12,'Oct'],[16,'Apr'],[19,'Feb'],[11,'Dec'],[9,'Jul']] 
```

```python
birthday_ddmm[0]    #get the birthday of the first person
```

```python
birthday_ddmm[0][1] #get birthday month of the first person
```

### [3.2] Methods

```python
name_list = ["Leo", "Nora", "Emma", "James", "Lucy"]
```

- **append**: Add an element to the end of the list.

```python
name_list.append("Henry") 
name_list
```

- **insert**: Add an element at given position.

```python
name_list.insert(2,'Mia') 
name_list
```

- **remove**: Remove the first occurrence of the element.

```python
numbers = [10, 20, 30, 30, 30, 40, 50]
```

```python
numbers.remove(20)  
numbers
```

```python
numbers.remove(30)  
numbers
```

- **pop**: Remove and return the element at the given index, or the last element if no index is given.

```python
mylist = ['a', 'b', 'c', 'd', 'e']
mylist.pop(1)
```

```python
print(mylist)
```

```python
mylist = ['a', 'b', 'c', 'd', 'e']
mylist.pop()
```

```python
print(mylist)
```

## [4] Lists and strings

- **Break a string into words**

```python
mystr = 'Python for Everybody'
mystr.split()
```

- **Split a string by seperator**

```python
mystr = 'Python-for-Everybody'
mystr.split('-')     # mystr.split(sep = '-')
```

- **Concatenates the elements by join method**

```python
mylist = ['Python', 'for', 'Everybody']
delimiter1 = ' '
delimiter1.join(mylist)
```

```python
delimiter2 = '-'
delimiter2.join(mylist)
```

- **Mutable and immutable<br>**

```python
# Lists are mutable
list_1 = [1, 2, 3]
list_1.append(4)
print(list_1)
```

```python
# strings are immutable
str_1 = "ABC"
str_1.replace("A","Z")
```

```python
# The previous step "replace" did not change the value of str1
str_1
```

```python
str_2 = str_1.replace("A","Z")
str_2
```
