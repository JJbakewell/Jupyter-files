## [2] While loop

### [2.1] Updating variables

- **<code>while</code>**

```python
n = 5
while n > 0:
    print(n)
    n = n - 1
```

### [2.2] Break & continue

- **break**

```python
n = 0
while n < 10:
    n = n + 1
    if n == 5:
        break   
    print(n) 
```

- **continue**

```python
n = 0
while n < 10:
    n = n + 1
    if n == 5:
        continue 
    print(n)    
```

## Exercise.B

**(B.1) You have a part-time job that pays 300 kr per day. Your goal is to earn at least 2500 kr. However, your work schedule follows a specific pattern: you work for two consecutive days, then take the third day off, and continue this pattern until your total income meets or exceeds 2500 kr. Write a program using a while loop to calculate and print your cumulative income for each working day. Continue the loop until your total income reaches or exceeds your goal of 2500 kr.**<br>

Expected output:<br>
Day-1: 300<br>
Day-2: 600<br>
Day-4: 900<br>
Day-5: 1200<br>
...<br>
Day-13: 2700<br>

```python
day = 0
income = 0
while income < 2500:
    day = day + 1
    
    if day % 3 == 0:
        continue
    else:
        income = income + 300
    print(f"day-{day}:{income}")
```

## [3] For loop

### [3.1] Iterable objects - string, list and range

```python
mystr = "python"
for i in mystr:
    print(i)
```

### [3.2] Break & continue

### [3.3] Applications

- **Summation**

```python
# Each item represents the number of confirmed cases per day
covid_list = [10, 15, 12, 10, 18] 

# Initialise a variable (cumulative number of cases)
total_cases = 0

# For each iteration, we update the variable "total_cases"
for daily_cases in covid_list:
    total_cases = total_cases + daily_cases
    print(total_cases)
```

- **Counter**

```python
animal_list = ['dog', 'dog', 'cat', 'dog', 'cat', 'cat', 'dog', 'cat', 'dog', 'cat']

# Initialise a variable
counter = 0

# For each iteration, we update the variable "counter"
for animal in animal_list:
    if animal == 'dog':
        counter = counter +1
    
print('# of dogs:', counter)
```

- **Find the maximum number**

```python
price_list = [100, 450, 200, 250, 300, 500, 150]

# Initialise a variable
max_price = price_list[0]


# For each iteration, if the test is true, we update the variable "max_price"
for price in price_list:
    if price > max_price:
        max_price = price

print("Maximum:", max_price) 
```
