# (1) While loop

**(1.1) Write a while loop to print out the following sequence of numbers on one line.**<br>
Expected result:<br>
<code>3 6 9 12 15 18 21 24 27 30 </code>

```python
n = 3
while n <= 30:
    print(n, end = " ")
    n = n + 3
```

Write a program using a while loop to remove the last item from the list until the total weight of the items in your list is below or equal to 23 kg.
```python
items = [5, 3, 6, 3, 4, 8, 2, 5, 4, 8, 3]
```

**(2.2) Write a program that asks the user to enter a positive integer. Calculate the factorial of this number.**<br> 
Hint: The factorial of a positive integer $n$ is represented as $n!$ and is calculated as the product of all positive integers from $1$ to $n$. For example, the factorial of 6 is $6! = 1\times2\times3\times 4\times5\times6 = 720$.<br> 


Example-1:<br> 
Enter a positive number:<code>6</code>   
The factorial of 6 is 720.


Example-2:<br> 
Enter a positive number:<code>4</code>   
The factorial of 4 is 24.

```python
n = int(input("Enter a positive number: "))
factorial = 1  
for i in range(1, n + 1):
    factorial*= i
    
print(f"The factorial of {n} is {factorial}.")
```

**(2.4) Given the list below that represents your daily expenses for 10 days. Assuming you have an initial balance of 3000 on day 0, write a program to calculate and print your balance for each day from day 1 to day 10.** (The balance at the end of day 1 is 3000-340 = 2660.)<br>

Expected result:<br>
2660<br>
2430<br>
...<br>
...<br>
560

```python
spend = [340, 230, 150, 180, 350, 110, 490, 210, 260, 120]
```

```python
balance = 3000
for s in spend:
    balance -= s
    print(balance)
```

**(2.6) Given the following list of sublists, where each sublist contains a customer's name and age, write a program to count the number of customers whose ages are between 20 and 25 (inclusive).**

```python
customer_list = [["Henry",28],["Victoria",33], ["Issac",19],["Maya",20],["Zoe",25],
                ["Nora",21],["Madelyn",29], ["Michael",35],["Mathew",23],["Sophia", 18]]
```

```python
counter = 0
for c in customer_list:
    if 20 <= c[1] <= 25:
        counter += 1
        
print(counter)
```

**(2.7) Given a list of city names, identify and print the longest city name in the list.**

```python
city_list = ["Berlin", "Barcelona", "Tokyo", "Istanbul", "Bordeaux", "Copenhagen", "Bangkok", "Houston", "Melbourne"]
```

```python
longest_name = city_list[0]  

for city in city_list:
    if len(city) > len(longest_name):
        longest_name = city
        
print(longest_name)
```

**(2.8) A company stores personal information in the variable <code>info</code> as shown below. Convert the information into a masked version, where all digits in the text are replaced with hash symbols (<code>#</code>).** <br> 

Expected result:<br>
Name: Jane Smith <br>
DOB: ##/##/#### <br>
SSN: ###-##-#### <br>
Email: j.smith@gmail.com<br>

```python
info = "Name: Jane Smith \nDOB: 01/10/1986 \nSSN: 123-45-6789 \nEmail: j.smith@gmail.com"
print(info)
```

```python
masked_info = ""
for s in info:
    if s.isdigit():
        masked_info += "#"
    else:
        masked_info += s
print(masked_info)
```
