### [1] Nested loops
A nested loop is a loop inside a loop.


```python
list_1 = ["A", "B", "C"]
list_2 = [1, 2, 3, 4]

for i in list_1:
    for j in list_2:
        print(i, j)
```

```python
for i in range(1,4):
    for j in range(1,4):
        print (f"{i}-{j}")
```

```python
# Start a new line after each iteration of the outer for loop.
for i in range(1,4):
    for j in range(1,4):
        print (f"{i}-{j}", end = " ")
    print()  #same as print(end = "\n")
```

### [2] Enumerate function
- <code>enumerate()</code> is a function that allows you to track both the index (or position) and the value of each item during iteration in a loop.


```python
name_list = ["Leo", "Nora", "Emma", "James", "Lucy"]
```

```python
for index, value in enumerate(name_list):
    print(index, value)
```

- Use <code>start</code> to specify the starting value of the index counter.


```python
for index, value in enumerate(name_list, start = 1):
    print(index, value)
```

### [3] List comprehension

List comprehension is used to create new lists based on the values of an existing list.

- Example-1: Create a list of squares for numbers from 1 to 5.

```python
squares = [i**2 for i in range(1, 6)]

print(squares)
```

- Example-2: Convert temperature from Celsius to Fahrenheit.

```python
celsius = [20, 18, 33, 12, 8, 24, 17]

fahrenheit = [c*9/5+32 for c in celsius]

print(fahrenheit)
```

- Example-3: Convert the names to uppercase. 

```python
name_list = ["Leo", "Nora", "Emma", "James", "Lucy"]

name_upper = [name.upper() for name in name_list]

print(name_upper)
```

```python
scores = [10, 7, 5, 9, 4, 6, 8, 9, 8]

pass_list = [ s for s in scores if s > 6.5 ]

print(pass_list)
```

- Determine the value of the new list based on conditions.


```python
scores = [10, 7, 5, 9, 4, 6, 8, 9, 8]

results = [ "pass" if s > 6.5 else "fail" for s in scores]

print(results)
```
