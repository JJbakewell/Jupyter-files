**(1) Suppose you are rolling a pair of dice. Print out all possible results.**<br>
Expected result:<br>
1&1 1&2 1&3 1&4 1&5 1&6 <br>
2&1 2&2 2&3 2&4 2&5 2&6 <br>
3&1 3&2 3&3 3&4 3&5 3&6 <br>
4&1 4&2 4&3 4&4 4&5 4&6 <br>
5&1 5&2 5&3 5&4 5&5 5&6 <br>
6&1 6&2 6&3 6&4 6&5 6&6

```python
for i in range(1,7):
    for j in range(1,7):
        print(f"{i}&{j}", end = " ") 
    print()
```

**(2) Write a nested for loop to print out the following pyramid pattern.**<br>
1 <br>
2 2 <br>
3 3 3 <br>
4 4 4 4 <br>
5 5 5 5 5 <br>
6 6 6 6 6 6 <br>

```python
for i in range(1,7):
    for j in range(i):
        print(i, end = " ")
    print()
```

**(3) Given the list below that represents your daily expenses for 10 days. Assuming you have an initial balance of 3000 on day 0, write a program to calculate and print your balance at the end of each day from day 1 to day 10.** (Use the enumerate() function to indicate the day in your output.)<br>

Expected result:<br>
Day1: 2660<br>
Day2: 2430<br>
...<br>
...<br>
Day10: 560

```python
spend = [340, 230, 150, 180, 350, 110, 490, 210, 260, 120]
```

```python
balance = 3000
for index, value in enumerate(spend, 1):
    balance -= value
    print(f"Day{index}: {balance}")
```

**(4) Write a for loop to go through all grades and print the Student_ID for each student who got an 'A'.** <br>
Expected output:<br>

|Student_ID|Grade|
|--:|--:|
|1|D|
|2|A|
|3|B|
|4|C|
|5|A|
|6|B|
|7|C|

```python
grades = ["D", "A", "B", "C", "A", "B", "C"]
```

**(8) Students took two exams, and their scores are stored in the <code>scores</code> list as pairs. The final score for each student is determined by taking the higher of the two exam scores. Write a program to calculate the final score for each student and store the results in a new list.** 

|Student|Scores (exam1, exam2)|Final score|
|:-:|:-:|:-:|
|1|7, **8**|8|
|2|**9**, 5|9|
|3|6, **7**|7|
|4|**6**, 5|6|
|5|**8**, 6|8|


```python
scores = [[7, 8], [9, 5], [6, 7], [6, 6], [8, 6]]
```

```python
final_scores = [max(i) for i in scores]
final_scores
```

**(9) Given the following list, which contains sublists with each customer's name and age, identify the customers whose ages are between 20 and 25 (inclusive) and store their details in a new list.**<br>
Expected result: <br>
<code>[['Maya', 20], ['Zoe', 25], ['Nora', 21], ['Mathew', 23]]</code>

```python
customer_list = [["Henry",28],["Victoria",33], ["Issac",19],["Maya",20],["Zoe",25],
                ["Nora",21],["Madelyn",29], ["Michael",35],["Mathew",23],["Sophia", 18]]
```

```python
filtered_customers = [customer for customer in customer_list if 20 <= customer[1] <= 25]
filtered_customers
```
