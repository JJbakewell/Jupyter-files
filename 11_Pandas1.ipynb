## [2] Pandas data structures
- **Series**: A one-dimensional array-like structure.
- **DataFrame**: A tabular, spreadsheet-like structure.

### [2.1] Series

- **Create a series from a list**<br>

Example: Jan–Jun temperatures (°C)

````python
temperature = pd.Series([-3, -3, 2, 5, 12, 16])
temperature
````

````python
# Value: The actual data inside the series
temperature.values
````

````python
# Index: The labels that identify each value
temperature.index
````

- **Create a series and assign an index array**

````python
temperature = pd.Series([-3, -3, 2, 5, 12, 16], index = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"])
temperature
````

````python
temperature.index
````

````python
temperature["May"]
````

### [2.2] DataFrame

- **Create a DataFrame from a dictionary**<br>

Example: 2021–2024 Average Monthly Salary (in 1,000 EUR)

````python
# Create a dictionary
salary_data = {"Year": ["2021", "2022", "2023", "2024"],
               "Norway": [5.2, 5.3, 5.4, 5.5],
               "Sweden": [4.8, 4.9, 5.0, 5.1],
               "Denmark": [5.0, 5.1, 5.2, 5.3]}
````

````python
# Create a DataFrame from a dictionary
salary_df = pd.DataFrame(salary_data)
salary_df
````

- **Create a DataFrame and assign an index array**<br>

Example: 2020–2024 Unemployment Rate (%)

````python
unemployment_data = {"Norway": [4.6, 4.0, 3.4, 3.5, 3.7],
                     "Sweden": [8.3, 8.8, 7.5, 7.8, 7.6],
                     "Denmark": [5.6, 5.1, 4.8, 4.6, 4.5]}
````

````python
unemployment_df = pd.DataFrame(unemployment_data, index = ["2020", "2021", "2022", "2023", "2024"])
unemployment_df
````

````python
unemployment_df.shape
````

- **A DataFrame is a collection of Series**

````python
unemployment_no = unemployment_df["Norway"]

print(unemployment_no)
````

````python
type(unemployment_no)
````

### [3.2] Descriptive statistics

- **Pandas dtype**

|Pandas dtype|python built-in type|Description|
|:--|:--|:--|
|int64|int|Integer numbers|
|float64|float|Floating point numbers|
|object|str or mixed|Text or mixed numeric and non-numeric values|
|string|str|Text|
|bool|bool|True/False values|
|category|str|Finite list of text values|
|datetime64|datetime|Date and time values|
|timedelta[ns]|--|Differences between two datetimes|

- **You can set the column data types with <code>dtype</code> based on your domain knowledge.**
```Python
        pd.read_csv("file_name", dtype = {"column_name": <dtype>})
```

````python
park_df = pd.read_csv("parks.csv", dtype = {"Acres": float})
park_df.info()
````

- **Descriptive statistics of numeric columns**

````python
park_df.describe()
````

- **Value counts for a categorical column**

````python
park_df.State.value_counts()
````

# [4] Sorting

Example: 2022–2024 Inflation Rate (%)

````python
inflation_data = {"country": ["Denmark", "Norway", "Sweden","Denmark", "Norway", "Sweden","Denmark", "Norway", "Sweden"],
                     "year": [2022, 2022, 2022, 2023, 2023, 2023, 2024, 2024, 2024],
                     "inflation_rate": [7.7, 5.8, 8.4, 4.0, 4.8, 6.6, 2.5, 3.5, 4.0]}

inflation_df = pd.DataFrame(inflation_data)
inflation_df
````

- **Sort a DataFrame by one column**

````python
inflation_df.sort_values(by = "inflation_rate")
````

- **Sort a DataFrame in descending order**

````python
inflation_df.sort_values(by = "inflation_rate", ascending = False)
````

- **Sort the dataframe and store the result**<br>

Most DataFrame methods are NOT in-place methods by default, so they:
- leave the original DataFrame unchanged
- return a new DataFrame



````python
# Assign the result to a new variable
df_sorted = inflation_df.sort_values(by = "inflation_rate")
df_sorted
````
